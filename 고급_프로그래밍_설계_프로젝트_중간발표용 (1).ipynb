```python
import pandas as pd
from pandas import DataFrame

df = pd.read_csv("2025년+전국극장조사.csv")
df
```

```python
#모집단인 대기업 영화관의 총 스크린 수, 총 좌석수, 특별관명, 특별관 스크린 수 추출

cinema_df =df[["영화관명", "개관일", "총 스크린 수", "총 좌석수", "특별관명", "특별관 스크린수"]]
# '영화관명' 컬럼에서 'CGV', '메가', '롯데' 중 하나라도 포함하는 행을 필터링합니다.
# `.str.contains()` 메서드는 Series의 각 문자열에 대해 부분 문자열 포함 여부를 확인합니다.
# 'CGV|메가|롯데'는 정규 표현식으로 'CGV' 또는 '메가' 또는 '롯데'를 의미합니다.
# na=False는 NaN 값을 포함하지 않도록 처리합니다.
revised_cinema_df = cinema_df[df["영화관명"].str.contains('CGV|메가|롯데', na=False)]

# NaN 값을 가진 행을 revised_cinema_df에서 제거합니다.
revised_cinema_df = revised_cinema_df.dropna()

revised_cinema_df
```

```python
import pandas as pd
import unicodedata
import re
#df_brand에서 영화관별 브랜드를 입력 받음 (CGV, 메가박스, 롯데시네마), replacement_dict 변수는 재정의 하여 대체할 특별관명  변수
def process_special_screens_for_brand(df_brand, replacements_dict):
    special_screen_counts = {}

    sorted_replacements = sorted(replacements_dict.items(), key=lambda item: len(item[0]), reverse=True)

    for index, row in df_brand.iterrows():
        special_names = str(row['특별관명']).strip()

        if special_names == '~' or special_names == 'nan' or not special_names:
            continue

        def apply_replacements(text):
            text = unicodedata.normalize('NFC', text)
            for k_term, e_term in sorted_replacements:
                text = text.replace(k_term, e_term)
            return text
    #각 극장(Individual_screen_entries) 변수의 특수관명은 각각 ','를 기준으로 분활함
        individual_screen_entries = [s.strip() for s in special_names.split(',')]
    #각 개별 항목에서 정규 표현식 r'(.+?)\((\d+)\)$'을 사용하여 특별관이름(숫자) 형식(Comfort(5))으로 되어 있는 경우 특별관 이름과 개수를 추출, 해당 형식이 아니면 개수를 1로 간주
        for entry in individual_screen_entries:
            match = re.match(r'(.+?)\((\d+)\)$', entry)
            if match:
                screen_type_raw = match.group(1).strip()
                count = int(match.group(2))
            else:
                screen_type_raw = entry.strip()
                count = 1

            screen_type = apply_replacements(screen_type_raw)
            special_screen_counts[screen_type] = special_screen_counts.get(screen_type, 0) + count

    # CSV에서 확인한 특수관명 목록에서 (~)물결표는 사실상  결측값 이므로 del명령어를 통해 집계 항목에서 제외(삭제) 시킴
    if '~' in special_screen_counts:
        del special_screen_counts['~']

    # Convert to Series and sort
    return pd.Series(special_screen_counts).sort_values(ascending=False)

def clean_and_process_brand_df(df_brand, brand_name, replacements_dict):
    # Ensure '총 스크린 수' and '총 좌석수' are numeric
    for col in ['총 스크린 수', '총 좌석수']:
        df_brand[col] = df_brand[col].astype(str).str.replace(',', '', regex=False)
        df_brand[col] = pd.to_numeric(df_brand[col].replace('~', pd.NA), errors='coerce')

    # Convert '개관일' to datetime
    df_brand['개관일'] = pd.to_datetime(df_brand['개관일'], errors='coerce')
    df_brand.dropna(subset=['개관일'], inplace=True)

    # Process special screens
    special_screen_series = process_special_screens_for_brand(df_brand, replacements_dict)
    print(f"\n{brand_name} Special Screen Counts:")
    print(special_screen_series)

    # Calculate yearly new openings (2019-2025)
    df_brand['year'] = df_brand['개관일'].dt.year
    new_openings = df_brand[(df_brand['year'] >= 2019) & (df_brand['year'] <= 2025)]
    new_openings_yearly = new_openings.groupby('year').size().reset_index(name='New Openings')
    new_openings_yearly.rename(columns={'year': 'Year'}, inplace=True)

    print(f"\n{brand_name} Yearly New Openings (2019-2025):")
    print(new_openings_yearly)
    return special_screen_series, new_openings_yearly

# 특별관명 변수들 재정의
common_replacements = {
    '4DX SCREEN': '4DX', 'MX 4D': 'MX4D', 'SUPER PLEX': 'Superplex', 'SUPER S': 'Super S',
    'Screen X': 'ScreenX', '수퍼 4D': 'Super 4D', '컴포트리클라이너': 'recliner', '리클라이너': 'recliner',
    '컴포트 시트': 'Comfort', '컴포트': 'Comfort', '더 부티크 스위트': 'The Boutique Suite',
    '더 부티크': 'The Boutique', 'LED 스크린': 'LED Screen', '탬퍼 시네마': 'Tempur Cinema',
    '골드클래스': 'Gold Class', '시네 드 쉐프': 'Cine de Chef', 'Cine de chef': 'Cine de Chef',
    '프라이빗 박스': 'Private Box', '시네 키즈': 'Cine Kids', '수퍼플렉스': 'Superplex',
    '돌비 애트모스': 'Dolby Atmos', '광음시네마': 'Kwangum', 'kwangum': 'Kwangum',
    '광음': 'Kwangum', '돌비 시네마': 'Dolby Cinema', '수퍼 LED': 'LED Screen',
    '슈퍼플렉스': 'Superplex', '슈퍼 4D': 'Super 4D', 'SCREENX': 'ScreenX'
}


# --- CGV 특별관명 변수 재정의---
print("--- CGV Analysis ---")
cgv_df = revised_cinema_df[revised_cinema_df["영화관명"].str.contains('CGV', na=False)].copy()
cgv_replacements = common_replacements.copy()
cgv_replacements.update({
    'ScreenX+리클': 'ScreenX+recliner',
    'ScreenX리클+PB': 'ScreenX+recliner+PB', #여기서 PB는 private box를 의미함
    'Cine de chef': 'Cine de Chef', # CSV에서 수집자의 휴먼에로 인해 중복으로 입력된 변수 Cine de Chef와 cine de chef를 통합시킴  (집계시 혼선을 없애기 위해 대,소문자 차이 없이 하나로 집계하도록 설정)
    '리클+아트하우스': 'recliner+Art House',
    '라운지체어': 'Lounge Chair',
    '돌비 애트모스 리클라이너': 'Dolby Atmos+recliner',
    'IMAX': 'IMAX',
    'SphereX': 'SphereX',
    '스위트 시네마': 'Suite Cinema',
    'Cine & Livingroom': 'Cine & Living Room',
    '씨네 앤 포레': 'Cine & Foret',
    '모션베드': 'MotionBed',
    '헤드폰': 'Headphone',
    '리클라이너 전용관': 'recliner'
})
cgv_special_screen_series, cgv_new_openings_yearly = clean_and_process_brand_df(cgv_df, 'CGV', cgv_replacements)


# --- 메가변수 특별관명 ---
print("\n--- Megabox Analysis ---")
megabox_df = revised_cinema_df[revised_cinema_df["영화관명"].str.contains('메가', na=False)].copy()
megabox_replacements = common_replacements.copy()
megabox_replacements.update({
    'Dolby Atmos': 'Dolby Atmos',
    'The Boutique Suite': 'The Boutique Suite',
    'Dolby V+A': 'Dolby V+A',
    'Le Recliner': 'Le Recliner',
    'The Boutique': 'The Boutique',
    'Dolby Cinema': 'Dolby Cinema',
    'The Boutique Private1': 'The Boutique Private1',
    'The Boutique Private2': 'The Boutique Private2',
    'MX4D': 'MX4D',
    'Comfort(5)': 'Comfort',
    'Comfort(6)': 'Comfort',
    'Comfort(3)': 'Comfort',
    'MEGA LED Screen': 'MEGA LED Screen',
    'LED Screen': 'LED Screen',
    'DolbyAtmos': 'Dolby Atmos' # 위와 동일
})
megabox_special_screen_series, megabox_new_openings_yearly = clean_and_process_brand_df(megabox_df, 'Megabox', megabox_replacements)


# --- Lotte Cinema Analysis ---
print("\n--- Lotte Cinema Analysis ---")
lotte_df = revised_cinema_df[revised_cinema_df["영화관명"].str.contains('롯데', na=False)].copy()
lotte_replacements = common_replacements.copy()
lotte_replacements.update({
    'Charlotte': 'Charlotte',
    '리클라이너': 'recliner',
    'Super LED': 'LED Screen', # 동일명칭 통합
    'SuperLED': 'LED Screen', # 위와 동일
    'Super 4D': 'Super 4D',
    'Superplex': 'Superplex',
    '광음시네마': 'Kwangum',
    'MX 4D': 'MX4D', # 위와 동일하게 오타차이(띄어쓰기) 수정
    '광음시네마+리클': 'Kwangum+recliner',
    'Superplex G': 'Superplex',
    'Cine Couple': 'Cine Couple',
    'Cine Comfort': 'Comfort',
    'COLORIUM': 'Colorium',
    '씨네패밀리': 'Cine Family',
    '프리미엄': 'Premium'
})
lotte_special_screen_series, lotte_new_openings_yearly = clean_and_process_brand_df(lotte_df, 'Lotte Cinema', lotte_replacements)



```
